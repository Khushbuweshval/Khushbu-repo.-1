Controllling the flow in loop.

1.) Breaking statment :- Break the loop when an element is satisfying a given condition.

2.) Continue statment:- The loop continues the element in the given condition gets skipped.

```python
fruit=['Mango','Banana','pineapple','Guava','Grapes']
for i in fruit:
  if i=="Guava":# if the fruit become guava(4th iteration)
    break
  print(i)
```

```python
fruit=['Mango','Banana','pineapple','Guava','Grapes','potato','Tomato']
for i in fruit:
  if i=="Guava":# if the fruit become guava(4th iteration)
    continue
  print(i)
```

```python
#Arbitarary Argument , default type is tuple
def add (*args):
  print(type(args))
  print(args)
  sum=0
  for num in args:
    sum=sum+num
  return sum
```

```python
add(3,4)
```

```python
add(3,8,9,-3,-5,12)
```

```python
def func(a,b,*args):
  print(a)
  print(b)
  print(args)
```

```python
#Arbitary keyword arguments--**kwargs--- default type is dictionary
def display(**kwargs):
  print(type(kwargs))
  print(kwargs)

  for key, value in kwargs.items():
    print(key,"-->",value)
```
